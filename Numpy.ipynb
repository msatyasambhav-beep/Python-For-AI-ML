Question #2
given a postiive number n grater than 2 create a Numpy array of size (nxn) with all zeros and ones such that the ones make a shape of "+"

```python
n = 5
```

```python
import numpy as np
z = np.zeros((n,n), dtype='int')
print(z)
```

```python
#make middle row and column all 1
z[n//2,:] =1
z[:,n//2] =1

print(z)
```

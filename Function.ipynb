Even or Odd


```python
def even_odd(num):
  if num %2 == 0:
    print("Even")
  else:
    print("Odd")
```

```python
even_odd(100)
```

Factorial

```python
4! = 4*3*2*1
n! = n*n-1*n-2*....1
```

```python
def fact_number(n):
  fact = 1

  for i in range(1,n+1):
    fact = fact*i
  return fact
```

```python
fact_number(9)
```

Sum of all natural number till the number provided

```python
def sum_natural(n):
  num = 0

  for i in range(1,n+1):
    num = num+i
  return num
```

```python
sum_natural(23)
```

SET OPERATION

```python
A = {0,2,4,6,8}
B = {1,2,3,4,5}
```

```python
#Union()
print(A | B)
print(A.union(B))
print(len(A.union(B)))
```

```python
#Intersection
print(A & B)
print(A.intersection(B))
```

```python
#Difference
print(A - B)
print(A.difference(B))
```

```python
#Symmetric difference
print(A^B)
```

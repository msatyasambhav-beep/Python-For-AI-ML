**Map , Reduce & Filter**

```python
#Area of a circle
```

```python
import math

def area(r):
  return math.pi*(r**2)


radii = [1,2,3,4,5]


areas = []

for r in radii:
  a = area(r)
  areas.append(a)
```

```python
areas
```

```python
#### map(f, iterable)
```

```python
map(area, radii)
```

```python
list(map(area, radii))
```

Convert Celcius scale to farheneit

F = 9/5*C + 32

```python
temps = [("Mumbai",35), ("Berlin", 12), ("Tokyo",25 ), ("Hong Kong", 31), ("Sydney",30)]
```

```python
type(temps)
```

```python
len(temps)
```

```python
cel_to_f = lambda data: (data[0], (9/5)*data[1]+32)
```

```python
#map
map(cel_to_f,temps)
```

```python
list(map(cel_to_f,temps))
```

Filter Function - filters out the data

filter(f,data)

```python
import statistics

data = [1,2,3,4,5,9,12,4,2,9]
```

```python
avg = statistics.mean(data)
print(avg)
```

```python
filter(lambda x:x>avg, data)
```

```python
list(filter(lambda x:x>avg, data))
```

Reduce Function

reduce(f,data)



```python
from functools import reduce
```

```python
data = [1,2,3,4,5]
```

```python
multiplier = lambda x,y: x*y
```

```python
reduce(multiplier, data)
```

```python
data = [1,2,3,4,5]
```

```python
prev = data[0]

for i in data:
  i=i*prev
  prev = i

print(i)
```

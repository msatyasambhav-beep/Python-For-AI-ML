2. Deleting the rows with missing data

```python
updated_df = df.dropna(axis=0)
```

```python
updated_df.info
```

```python
3. Filling the Missing Values-Imputation
```

> Filling the missing data with the mean or median value if its a numerical variable.
> Filling the missing data with node its a categorical value
> filling the categorical value with a new type for the mssing values

```python
df["Age"].mean()
```

```python
df['Age'].median()
```

```python
updated_df['Age'] = updated_df['Age'].fillna(df['Age'].mean())
updated_df.info()
```

```python
##Iteration
```

```python
##iterate over a list of integer
list_1 = [10,20,30,40,50]
for i in list_1:
  print(i)
```

```python
#Iterator of String
str_1 = "India"
for i in str_1:
  print(i)
```

```python
#Iterator over a Dictionary

student_data = {1:["Sam",25],2:["Sharma",21],3:["Ravi",23],4:["SHalu",23],5:["Side",22]}
for i in student_data:
  print(i)
```

```python
#Iterator over a Dictionary

student_data = {1:["Sam",25],2:["Sharma",21],3:["Ravi",23],4:["SHalu",23],5:["Side",22]}
for i in student_data.items():
  print(i)
```

```python
#Iterate over a range of values
for i in range(1,5):
  print(i)
```

```python
list_1 = ["Automobiles","Honda","Benz","Maruti","Kia"]

list_2 = []

for i in list_1:
  list_2.append(i)

list_2 = [len(i) for i in list_1]
print(list_2)
```

```python
list_1 = ["Automobiles","Honda","Benz","Maruti","Kia"]

d = {i : len(i) for i in list_1}

d
```

```python
list_1 = ["Automobiles","Honda","Benz","Maruti","Kia"]

d={}

for i in list_1:
  d[i] = len(i)

d
```
